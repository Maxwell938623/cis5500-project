## Handle missing values and basic cleanup


```
# see what station coords have nan
print(stationCoordsDfRaw.shape)
currDf = stationCoordsDfRaw
currDf[~currDf.isna().any(axis=1)].head() # clean data
currDf[currDf.isna().any(axis=1)] # dirty data

# It seems like it's mostly ADA Notes that is NaN so let's remove that col and see
currDf = stationCoordsDfRaw.drop(columns=["ADA Notes"])
print(currDf.shape)
print(currDf[~currDf.isna().any(axis=1)].shape) # clean data
print(currDf[currDf.isna().any(axis=1)].shape) # dirty data

#Conclusion: ADA Notes not necessary and introduces NaN values so we get rid of that feature
```

```
#drop the only dirty column
stationCoordsDfClean = stationCoordsDfRaw.drop(columns=["ADA Notes"])
```

```
# clean Fair evasion stats

# we only need years [2020, 2024]
fairEvDfGoodYears = fairEvasionStatsDfRaw[(fairEvasionStatsDfRaw['Time Period'] >= "2020-Q1") & (fairEvasionStatsDfRaw['Time Period'] <= "2024-Q4")]
fairEvDfGoodYears[fairEvDfGoodYears.isna().any(axis=1)] # only 2020-Q2 is invalid at all (and it is completely invalid)

# We decide to handle 2020-Q3 case by case on querie (we will keep the NaN). Start of Covid too irregular to impute with simple methods
# We may try to impute later based on ridership data
fairEvasionStatsDfClean = fairEvDfGoodYears
```

```
# transit_timestamp entries are of the form    01/01/2020 12:00:00 AM
# we want year and quarter attributes so that joins with fair evasion statistic are easier
ridershipDfsWithAttrs = []
# note: running this cell changes ridershipDfsClean. these are large dataframes so not worth making copies
for i, df in enumerate(ridershipDfsClean):
  df['year'] = df['transit_timestamp'].str[6:10].astype(int)
  df['quarter'] =  np.ceil(df['transit_timestamp'].str[:2].astype(int) / 3).astype(int)
  ridershipDfsWithAttrs.append(df)
  print(df.shape, df.columns)
```

```
fairEvasionStatsDfWithAttrs = fairEvasionStatsDfClean.copy() # this is small
fairEvasionStatsDfWithAttrs['year'] = fairEvasionStatsDfWithAttrs['Time Period'].str[:4].astype(int)
fairEvasionStatsDfWithAttrs['quarter'] = fairEvasionStatsDfWithAttrs['Time Period'].str[-1].astype(int)
fairEvasionStatsDfWithAttrs.drop(columns=["Time Period"], inplace=True)
fairEvasionStatsDfWithAttrs.head()
```

## Derive attributes and normalize types


```
#georeference is a copy of latitude and longitude that are already attributes, so we drop georeference
stationCoordsDfWithAttrs = stationCoordsDfClean.drop(columns=['Georeference'])
print(stationCoordsDfWithAttrs.shape, stationCoordsDfWithAttrs.columns)
```

```
print("before:")
print(ridershipDfsWithAttrs[0].dtypes)
ridershipDfsWithAttrs[0].head()
# print([x for x in ridershipDfsWithAttrs[0]['transfers'].unique()])
#oddly it seems that ridership numbers are mostly strings like "1,337" or ints.
# and transfers does the same



# ints: ridership, transfers
# drop georeference
# categorical: station_complex, station_complex_id, transit_mode, borough, payment_method, fare_class_category

ridershipDfsFinal = []
for i, df in enumerate(ridershipDfsWithAttrs):
  df = df.drop(columns=["Georeference"])
  for intAttr in ("ridership", "transfers"):
    df[intAttr] = df[intAttr].astype("string").str.replace(",", "").astype(int)

  for catAttr in ("station_complex", "station_complex_id", "transit_mode", "borough", "payment_method", "fare_class_category"):
    df[catAttr] = df[catAttr].astype('category')
  ridershipDfsFinal.append(df)

print("\n\nAfter:")
print(ridershipDfsFinal[0].dtypes)
```

```
fairEvasionStatsDfWithAttrs.dtypes
fairEvasionStatsDfWithAttrs.head() #fair evasion looks like : 11.3%   margin of error looks like : 0.9%

fairEvasionStatsDfFinal = fairEvasionStatsDfWithAttrs.copy() # small
fairEvasionStatsDfFinal["Fare Evasion"] = fairEvasionStatsDfFinal["Fare Evasion"].str[:-1].astype('float32') / 100
fairEvasionStatsDfFinal["Margin of Error"] = fairEvasionStatsDfFinal["Margin of Error"].str[:-1].astype('float32') / 100

print(fairEvasionStatsDfFinal.dtypes)
```

```
stationCoordsDfWithAttrs.head()

#int: Station ID, Complex ID, ADA,
#bool: ADA Northbound, ADA Southbound, CBD (already bool)
#cat: GTFS Stop ID, Division, Line, Stop Name, Borough, Daytime Routes, Structure, North Direction Label, South Direction Label

stationCoordsDfFinal = stationCoordsDfWithAttrs.copy()
for intAttr in ("Station ID", "Complex ID", "ADA"):
  stationCoordsDfFinal[intAttr] = stationCoordsDfFinal[intAttr].astype(int)
for catAttr in ("GTFS Stop ID", "Division", "Line", "Stop Name", "Borough", "Daytime Routes", "Structure", "North Direction Label", "South Direction Label"):
  stationCoordsDfFinal[catAttr] = stationCoordsDfFinal[catAttr].astype('category')
for boolAttr in ('ADA Northbound', 'ADA Southbound'):
  stationCoordsDfFinal[boolAttr] = stationCoordsDfFinal[boolAttr].astype(bool)

print(stationCoordsDfFinal.dtypes)
```

## Save processed datasets


```
outputCSVPath = '/content/drive/My Drive/CIS 4500 HW/CIS4500FinalProject/ProcessedDataCSVs'
#save CSVs to drive

for i, ridershipDf in enumerate(ridershipDfsFinal):
  ridershipDf.to_csv(os.path.join(outputCSVPath, "ridership" + str(i + 2020) + "Dataframe.csv"), index = False)
stationCoordsDfFinal.to_csv(os.path.join(outputCSVPath, "stationCoordsDataframe.csv"), index = False)
fairEvasionStatsDfFinal.to_csv(os.path.join(outputCSVPath, "fairEvasionStatsDataframe.csv"), index = False)
```
