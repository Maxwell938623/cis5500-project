## Clean arrests records


```python
import pandas as pd
import numpy as np
import os

csvsFolderPath = '/content/drive/My Drive/CIS 4500 HW/CIS4500FinalProject/RawDataCSVs'
nypdArrestsPath = os.path.join(csvsFolderPath, "nypd_arrests.csv")

df = pd.read_csv(nypdArrestsPath)
print(df.shape)
print(df.dtypes)
```

```python
df.head()
```

```python
print(df[df.isna().any(axis=1)].shape)  # how many dirty rows?
df.isnull().sum()  # NaN count per column
```

```python
df_clean = df.dropna(subset=['Latitude', 'Longitude'])
print(df_clean.shape)
```

```python
df_clean = df_clean.drop(columns=['Lon_Lat'])
```

```python
print(df_clean.shape)
```

```python
df_clean = df_clean.drop(columns=['X_COORD_CD', 'Y_COORD_CD'])
```

```python
print(df_clean.shape)
```

```python
df_clean['ARREST_DATE'] = pd.to_datetime(df_clean['ARREST_DATE'])
df_clean['year'] = df_clean['ARREST_DATE'].dt.year
df_clean['month'] = df_clean['ARREST_DATE'].dt.month
df_clean['quarter'] = df_clean['ARREST_DATE'].dt.quarter
print(df_clean.shape)
```

```python
# Numeric IDs — keep as int
for col in ['PD_CD', 'KY_CD', 'ARREST_PRECINCT', 'JURISDICTION_CODE']:
    df_clean[col] = pd.to_numeric(df_clean[col], errors='coerce').astype('Int64')  # Int64 handles NaN

# Categorical columns (low-cardinality text)
for col in ['LAW_CAT_CD', 'ARREST_BORO', 'AGE_GROUP', 'PERP_SEX', 'PERP_RACE',
            'PD_DESC', 'OFNS_DESC', 'LAW_CODE']:
    df_clean[col] = df_clean[col].astype('category')

# ARREST_KEY stays as string (it's an ID)
df_clean['ARREST_KEY'] = df_clean['ARREST_KEY'].astype('string')

print(df_clean.dtypes)
```

```python
df_clean = df_clean.drop(columns=['KY_CD', 'OFNS_DESC', 'LAW_CODE', 'ARREST_BORO', 'ARREST_PRECINCT', 'quarter'])
```

```python
df_clean.shape
```

```python
df_clean = df_clean[(df_clean['year'] >= 2020) & (df_clean['year'] <= 2024)]
print(df_clean.shape)
print(df_clean['year'].value_counts().sort_index())
```

```python
df_clean = df_clean[df_clean['JURISDICTION_CODE'] == 1]
print(df_clean.shape)
print(df_clean['year'].value_counts().sort_index())
```

```python
df_head = df_clean.head(100)
```

```python
# load in robertos mid
csvsFolderPath = '/content/drive/My Drive/CIS 4500 HW/CIS4500FinalProject/ProcessedDataCSVs'
robertosMidPath = os.path.join(csvsFolderPath, "robertosmid.csv")
robertosMidDf = pd.read_csv(robertosMidPath)
print(robertosMidDf.shape)
```

```python
from scipy.spatial import cKDTree

# Step 1: Handle duplicate station complex IDs by averaging their lat/lon
station_coords = robertosMidDf.groupby('station_complex_id')[['latitude', 'longitude']].mean().reset_index()
print(station_coords.shape)  # sanity check
station_coords.head()
```

## Map arrests to nearest station


```python
# Step 2: Build a KD-tree from station coordinates (super fast nearest neighbor lookup)
station_points = station_coords[['latitude', 'longitude']].values
tree = cKDTree(station_points)

# Step 3: Query the tree for every row in df_transit at once (vectorized, no loop!)
transit_points = df_clean[['Latitude', 'Longitude']].values
distances, indices = tree.query(transit_points)

# Step 4: Map the index back to station_complex_id
df_clean = df_clean.copy()  # avoid SettingWithCopyWarning
df_clean['station_complex_id'] = station_coords['station_complex_id'].iloc[indices].values

df_clean.head(10)
```

```python
# Sanity check — spot check a few rows manually
sample = df_clean[['Latitude', 'Longitude', 'station_complex_id']].sample(5)
for _, row in sample.iterrows():
    match = station_coords[station_coords['station_complex_id'] == row['station_complex_id']]
    print(f"Arrest @ ({row['Latitude']:.4f}, {row['Longitude']:.4f}) → station {row['station_complex_id']} @ ({match['latitude'].values[0]:.4f}, {match['longitude'].values[0]:.4f})")
```

```python
df_clean['ARREST_KEY'] = df_clean['ARREST_KEY'].astype('int64')
df_clean['station_complex_id'] = df_clean['station_complex_id'].astype('string')

df_clean = df_clean.drop(columns=['ARREST_DATE'])

print(df_clean.dtypes)
```
